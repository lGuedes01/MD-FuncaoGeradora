```python
# Multiplicação de polinômios usando listas.
#
# Um polinômio é guardado como uma lista de coeficientes, em que
# a posição na lista é o expoente de X:
#
#   1 + 2X + 3X^2   ->   [1, 2, 3]
#   1 + X^3         ->   [1, 0, 0, 1]
```

```python

def mult(p, q):
    # O produto tem grau (len(p)-1) + (len(q)-1),
    # então precisa de len(p) + len(q) - 1 coeficientes.
    resultado = [0] * (len(p) + len(q) - 1)

    # Cada termo de p é multiplicado por cada termo de q.
    # X^i vezes X^j dá X^(i+j), então o produto vai para a posição i+j.
    for i in range(len(p)):
        for j in range(len(q)):
            resultado[i + j] = resultado[i + j] + p[i] * q[j]

    return resultado
```

```python
def coef(p, n):
    # Coeficiente de X^n em p.
    # Potências além do fim da lista têm coeficiente 0.
    if n < len(p):
        return p[n]
    return 0
```

```python
def texto(p):
    # Transforma a lista de coeficientes em texto legível, ex. "(1 + 2X + X^2)"
    termos = []
    for expoente in range(len(p)):
        c = p[expoente]
        if c == 0:
            continue
        if expoente == 0:
            termos.append(str(c))
        elif expoente == 1:
            termos.append(str(c) + "X")
        else:
            termos.append(str(c) + "X^" + str(expoente))
    if len(termos) == 0:
        return "(0)"
    return "(" + " + ".join(termos) + ")"
```

### Questão 2:
- Sejam A(X) = a_0 + a_1 X + a_2 X^2 + ... e B(X) = b_0 + b_1 X + b_2 X^2 + ... os dois polinômios. Na coluna de X^k aparecem todos os termos a_i b_j tais que i + j = k, pois (a_i X^i) * (b_j X^j) = a_i b_j X^(i+j), e esse termo cai na coluna do expoente i + j.
- Na função mult, p[i] é o a_i e q[j] é o b_j. A linha resultado[i + j] = resultado[i + j] + p[i] * q[j] faz exatamente isso: resultado[i + j] é a coluna de X^(i+j), o resultado[i + j] acumula a soma da coluna, e p[i] * q[j] é um dos termos a_i b_j que pertencem a ela. Os dois laços percorrem todos os pares (i, j), então cada coluna recebe todos os seus termos. Por exemplo: em (1 + X)(1 + X + X^2), a coluna de X^2 recebe os termos a_0 b_2 e a_1 b_1, ou seja, 11 + 11 = 2.

### Questão 3:
- A lista "resultado" na função mult tem tamanho = len(p) + len(q) - 1, pois é necessário que o polinômio, após a multiplicação tenha como maior grau a soma dos dois maiores graus dos dois polinômios, no entanto, a lista tem sempre 1 de tamanho a mais do que o número de maior grau (e.g. se o polinômio é de grau 2, então len(p) = 3). Portanto, é necessário que faça a soma dos tamanhos menos 1 justamente para compensar o tamanho "extra" obtido com a soma do tamanho de mais um polinômio.

### Questão 4:

```python
def potencia(p,n):
    r = [1]
    for i in range(n):
        r = mult(p,r)
    return r
```

### Questão 13:

```python
def passo(passo, N):
    l= []
    for i in range(N+1):
        if i % passo == 0:
            l.append(1)
        else:
            l.append(0)
    return l
```

### Questão 14:

```python
def trunca(p, N):
    return p[:N + 1]

print(trunca([1, 2, 3, 4, 5], 2))
print(trunca([1, 2], 5))
```

- `p[:N+1]` Usando a dica, ela pega os índices 0, 1, ..., N, ou seja, só os coeficientes de grau <= N. Se p for menor que N+1, a fatia devolve a lista inteira.
